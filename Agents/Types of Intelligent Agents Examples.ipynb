This notebook demonstrates five common intelligent-agent types using a simple **smart room** example.

**Progression: React → Remember → Plan → Optimise → Learn**


## 1. Simple Reflex Agent

A **Simple Reflex Agent** acts only on the **current percept**. It has no memory.

**Rule:** IF condition → THEN action.

Here, the thermostat reacts to the current temperature.


```python
temperature = 18

if temperature < 22:
    action = "Turn heater ON"
else:
    action = "Turn heater OFF"

print("Current temperature:", temperature, "°C")
print("Agent action:", action)
```

```python
# Try several environmental conditions
temperatures = [18, 20, 22, 24, 26]

for temperature in temperatures:
    if temperature < 22:
        action = "Turn heater ON"
    else:
        action = "Turn heater OFF"
    print(temperature, "°C →", action)
```

## 2. Model-Based Reflex Agent

A **Model-Based Reflex Agent** uses the current percept plus an **internal state/model** of the environment.

**Current perception + Internal state → Action**

Here, the agent also knows whether the window is open.


```python
temperature = 18
window_open = True

if temperature < 22:
    if window_open:
        action = "Close the window"
    else:
        action = "Turn heater ON"
else:
    action = "Turn heater OFF"

print("Temperature:", temperature, "°C")
print("Window open:", window_open)
print("Agent action:", action)
```

## 3. Goal-Based Agent

A **Goal-Based Agent** chooses actions that move it towards a desired state.

Its question is:

**What action will help me reach my goal?**

Here, the goal is to reach 22°C.


```python
temperature = 18
goal = 22

print("Starting temperature:", temperature, "°C")
print("Goal:", goal, "°C")

while temperature < goal:
    print("Temperature:", temperature, "°C → Heat the room")
    temperature += 1

print("Temperature:", temperature, "°C")
print("Goal achieved!")
```

## 4. Utility-Based Agent

A **Utility-Based Agent** compares possible actions and selects the most desirable one.

Utility can represent comfort, cost, energy efficiency, safety, or other preferences.

Its question is:

**Which action gives the best outcome?**


```python
actions = {
    "Turn on air conditioner": 7,
    "Open window": 9,
    "Turn on fan": 6
}

for action, utility in actions.items():
    print(action, "→ Utility:", utility)

best_action = max(actions, key=actions.get)

print("\nBest action:", best_action)
print("Utility:", actions[best_action])
```

### Utility using two factors

The following example combines **comfort** and **energy efficiency** with equal importance.


```python
comfort = {
    "Air conditioner": 10,
    "Open window": 8,
    "Fan": 7
}

energy_efficiency = {
    "Air conditioner": 4,
    "Open window": 10,
    "Fan": 7
}

utility = {}

for action in comfort:
    utility[action] = (comfort[action] + energy_efficiency[action]) / 2

for action, score in utility.items():
    print(action, "→ Utility:", score)

best_action = max(utility, key=utility.get)
print("\nBest action:", best_action)
```

## 5. Learning Agent

A **Learning Agent** improves or adapts its behaviour using experience or data.

Here, the agent learns the user's preferred room temperature from previous selections.


```python
preferred_temperatures = [22, 23, 22, 24, 23]

learned_temperature = sum(preferred_temperatures) / len(preferred_temperatures)

print("Previous preferences:", preferred_temperatures)
print("Learned preferred temperature:", learned_temperature, "°C")
```

```python
current_temperature = 20

if current_temperature < learned_temperature:
    action = "Turn heater ON"
elif current_temperature > learned_temperature:
    action = "Turn heater OFF"
else:
    action = "Maintain current setting"

print("Current temperature:", current_temperature, "°C")
print("Learned target:", learned_temperature, "°C")
print("Agent action:", action)
```

### Learning from new experience

As the user provides more observations, the agent can update what it has learned.


```python
new_preferences = [24, 24, 23]

preferred_temperatures.extend(new_preferences)
learned_temperature = sum(preferred_temperatures) / len(preferred_temperatures)

print("Updated observations:", preferred_temperatures)
print("Updated learned temperature:", learned_temperature, "°C")
```

## Summary

| Agent Type | Main Idea | Easy way to remember |
|---|---|---|
| Simple Reflex | Current percept | **React** |
| Model-Based | Percept + internal state | **Remember** |
| Goal-Based | Desired future state | **Plan** |
| Utility-Based | Best outcome | **Optimise** |
| Learning | Experience/data | **Learn** |

> **Important:** These categories are not necessarily mutually exclusive. A sophisticated agent may maintain an internal model, pursue goals, optimise utility and learn from experience.
